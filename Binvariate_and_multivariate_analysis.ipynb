```python
import pandas as pd
import numpy as np
import gc
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"datasets\train_merged.csv")
```

```python
pd.crosstab(
    df['CODE_GENDER'],
    df['TARGET'],
    normalize='index'
)
```

```python
sns.countplot(
    data=df,
    x='CODE_GENDER',
    hue='TARGET'
)

plt.show()
```

```python
sns.boxplot(
    data=df,
    x='TARGET',
    y='AMT_INCOME_TOTAL'
)

plt.show()
```

```python
sns.histplot(
    data=df,
    x='AMT_INCOME_TOTAL',
    hue='TARGET',
    kde=True
)

plt.show()
```
