```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('datasets/train_merged.csv')
```

```python
df_clean = df.copy()
```

```python
df_clean["DAYS_EMPLOYED_ANOM"] = (
    df_clean["DAYS_EMPLOYED"] == 365243
).astype(int)

df_clean["DAYS_EMPLOYED"] = df_clean["DAYS_EMPLOYED"].replace(
    365243, np.nan
)
```

```python
df_clean
```

```python
df_clean["AGE_YEARS"] = (
    -df_clean["DAYS_BIRTH"] / 365.25
)
```

```python
df_clean["AGE_YEARS"].describe()
```

```python
df_clean["EMPLOYMENT_YEARS"] = (
    -df_clean["DAYS_EMPLOYED"] / 365.25
)
```

```python
df_clean["EMPLOYMENT_YEARS"].describe()
```

```python
missing_percentage = (
    df_clean.isnull().mean() * 100
).sort_values(ascending=False)

missing_percentage.head(50)
```

```python
high_missing_cols = missing_percentage[
    missing_percentage > 60
].index.tolist()

print("Columns with >60% missing:",
      len(high_missing_cols))
```

```python
for col in high_missing_cols[:50]:
    print(col, "->", missing_percentage[col])
```

```python
nunique = df_clean.nunique(dropna=False).sort_values()

nunique.head(30)
```

```python
constant_cols = nunique[nunique <= 1].index.tolist()

print("Constant columns:", len(constant_cols))
print(constant_cols)
```

```python
df_clean["EXT_SOURCE_1_MISSING"] = (
    df_clean["EXT_SOURCE_1"].isna()
).astype(int)
```

```python
missing_summary = pd.DataFrame({
    "missing_count": df_clean.isnull().sum(),
    "missing_percentage": df_clean.isnull().mean() * 100,
    "dtype": df_clean.dtypes
})

missing_summary["missing_category"] = pd.cut(
    missing_summary["missing_percentage"],
    bins=[-1, 0, 5, 20, 40, 60, 80, 90, 100],
    labels=[
        "0%",
        "<5%",
        "5-20%",
        "20-40%",
        "40-60%",
        "60-80%",
        "80-90%",
        "90-100%"
    ]
)

missing_summary["missing_category"].value_counts().sort_index()
```

```python
missing_summary.sort_values(
    "missing_percentage",
    ascending=False
).head(100)
```

```python
high_missing = missing_summary[
    missing_summary["missing_percentage"] > 60
].copy()

high_missing.shape
```

```python
high_missing["prefix"] = (
    high_missing.index
    .str.split("_")
    .str[0]
)

high_missing["prefix"].value_counts()
```

```python
# Remove extremely sparse columns
missing_percentage = df_clean.isnull().mean() * 100

drop_high_missing = missing_percentage[
    missing_percentage > 95
].index.tolist()

print("Columns to remove:", len(drop_high_missing))

df_clean = df_clean.drop(columns=drop_high_missing)

print("New shape:", df_clean.shape)
```

```python
print("Removed columns:")
for col in drop_high_missing:
    print(col)
```

```python
prev_cc_cols = [
    col for col in df_clean.columns
    if col.startswith("PREV_CC_")
]

print("PREV_CC columns remaining:", len(prev_cc_cols))
```

```python
prev_cc_missing = (
    df_clean[prev_cc_cols]
    .isnull()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

prev_cc_missing.head(20)
```

```python
missing_percentage = df_clean.isnull().mean() * 100

high_missing_80_95 = missing_percentage[
    (missing_percentage > 80) &
    (missing_percentage <= 95)
].index.tolist()

print("Columns with 80–95% missing:", len(high_missing_80_95))
```

```python
print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

remaining_missing = df_clean.isnull().sum().sum()

print("Total missing values:", remaining_missing)
```

```python
missing_percentage = df_clean.isnull().mean() * 100

high_missing_80_95 = missing_percentage[
    (missing_percentage > 80) &
    (missing_percentage <= 95)
].sort_values(ascending=False)

print("Number of columns:", len(high_missing_80_95))
print(high_missing_80_95)
```

```python
prev_cc_cols = [
    col for col in df_clean.columns
    if col.startswith("PREV_CC_")
]

print("PREV_CC columns:", len(prev_cc_cols))

prev_cc_missing = (
    df_clean[prev_cc_cols]
    .isnull()
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print(prev_cc_missing.head(20))
```

```python
# Missing-value summary after sparse-column removal

missing_summary = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": df_clean.isnull().mean() * 100
})

missing_summary = missing_summary[
    missing_summary["Missing_Count"] > 0
].sort_values("Missing_Percentage", ascending=False)

print("Columns still containing missing values:", len(missing_summary))
missing_summary.head(30)
```

```python
# Check how many applicants have any PREV_CC information
prev_cc_cols = [
    col for col in df_clean.columns
    if col.startswith("PREV_CC_")
]

prev_cc_missing_count = df_clean[prev_cc_cols].isna().all(axis=1)

print("PREV_CC columns:", len(prev_cc_cols))
print("Rows with ALL PREV_CC values missing:", prev_cc_missing_count.sum())
print(
    "Percentage:",
    prev_cc_missing_count.mean() * 100
)
```

```python
# Check whether PREV_CC missingness is related to TARGET

pd.crosstab(
    prev_cc_missing_count,
    df_clean["TARGET"],
    normalize="index"
) * 100
```

```python
df_clean["PREV_CC_ALL_MISSING"] = (
    df_clean[prev_cc_cols].isna().all(axis=1)
).astype(int)

print(
    df_clean["PREV_CC_ALL_MISSING"].value_counts()
)
```

```python
# Remove PREV_CC features derived from the applicant ID

prev_cc_id_cols = [
    col for col in df_clean.columns
    if col.startswith("PREV_CC_SK_ID_CURR_")
]

print("Columns to remove:", len(prev_cc_id_cols))

df_clean = df_clean.drop(columns=prev_cc_id_cols)

print("New shape:", df_clean.shape)
```

```python
print(
    "Remaining PREV_CC_SK_ID_CURR columns:",
    sum(col.startswith("PREV_CC_SK_ID_CURR_") for col in df_clean.columns)
)
```

```python
from pathlib import Path

# Create processed-data folder
processed_dir = Path("data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

# Save the CURRENT df_clean exactly as it is
output_path = processed_dir / "home_credit_cleaned_stage1.pkl"

df_clean.to_pickle(output_path)

print("Dataset saved successfully!")
print("Path:", output_path)
print("Shape:", df_clean.shape)
```
