```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('datasets/train_merged.csv')
```

```python
df_clean = df.copy()
```

```python
df_clean["DAYS_EMPLOYED_ANOM"] = (
    df_clean["DAYS_EMPLOYED"] == 365243
).astype(int)

df_clean["DAYS_EMPLOYED"] = df_clean["DAYS_EMPLOYED"].replace(
    365243, np.nan
)
```

```python
df_clean
```

```python
df_clean["AGE_YEARS"] = (
    -df_clean["DAYS_BIRTH"] / 365.25
)
```

```python
df_clean["AGE_YEARS"].describe()
```

```python
df_clean["EMPLOYMENT_YEARS"] = (
    -df_clean["DAYS_EMPLOYED"] / 365.25
)
```

```python
df_clean["EMPLOYMENT_YEARS"].describe()
```

```python
missing_percentage = (
    df_clean.isnull().mean() * 100
).sort_values(ascending=False)

missing_percentage.head(50)
```

```python
high_missing_cols = missing_percentage[
    missing_percentage > 60
].index.tolist()

print("Columns with >60% missing:",
      len(high_missing_cols))
```

```python
for col in high_missing_cols[:50]:
    print(col, "->", missing_percentage[col])
```

```python
nunique = df_clean.nunique(dropna=False).sort_values()

nunique.head(30)
```

```python
constant_cols = nunique[nunique <= 1].index.tolist()

print("Constant columns:", len(constant_cols))
print(constant_cols)
```
